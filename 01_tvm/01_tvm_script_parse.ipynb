# 01. 装饰器基础

```python
def my_decorator(func):
    def wrapper(*args, **kwargs):
        print(f"Calling function {func.__name__} with arguments: {args} and keyword arguments: {kwargs}")
        result = func(*args, **kwargs)
        print(f"Function {func.__name__} returned: {result}")
        return result
    return wrapper

def get_my_func():
    @my_decorator
    def my_func():
        print("Hello, world!")
    return my_func

my_func = get_my_func()
print(f"my_func run result: {my_func()}")
```

# 02. 装饰器逼近TL

```python
def another_func(func_v2):
    print(f"Calling function in another_func: {func_v2.__name__}")
    return func_v2

def my_parse(func_v2):
    return another_func(func_v2)
    
def my_decorator_v2(func_v2):
    def wrapper(func_v2):
        f = my_parse(func_v2)
        return f
    return wrapper(func_v2)
setattr(my_decorator_v2, "dispatch_token", "tir")

def get_my_func_v2():
    @my_decorator_v2
    def my_fun_v2():
        print("Hello, world! v2")
    return my_fun_v2

my_func_v2 = get_my_func_v2()
print(f"my_func_v2 run result: {my_func_v2()}")
```

# 03. TVM 解析 func 为 TIR

```python
from tvm.script.parser._core import parse, utils


def my_decorator_v3(func):
    return func

setattr(my_decorator_v3, "dispatch_token", "tir")

@my_decorator_v3
def my_fun_v3():
    print("Hello, world! v3")


warpped_func = parse(my_fun_v3, utils.inspect_function_capture(my_fun_v3))
print(warpped_func)
```
