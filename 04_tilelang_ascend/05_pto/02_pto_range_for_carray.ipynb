### 01 for循环：外层初始化，内层赋值，外层使用场景

```python
@pto.jit(target="a5", ast_rewrite=True)
def test1_last_value():
    m = pto.const(0, dtype=pto.i32)

    for i in range(4):
        m = pto.cast(i, pto.i32)
    _ = m

try:
    test1_last_value.compile().mlir_text()
    print("compiled successfully.")
except Exception as e:
    print(f"Error compiling: {e}")
```

### 02 for循环：外层初始化，内层引用赋值，外层使用场景

```python
@pto.jit(target="a5", ast_rewrite=True)
def test2_last_value():
    m = pto.const(0, dtype=pto.i32)

    for i in range(4):
        m = m + pto.cast(i, pto.i32)
    _ = m

try:
    test2_last_value.compile().mlir_text()
    print("compiled successfully.")
except Exception as e:
    print(f"Error compiling: {e}")
```

### 03 for循环：外层初始化，内层引用赋值，外层使用场景

```python
@pto.jit(target="a5", ast_rewrite=True)
def test3_last_value():
    m = pto.const(0, dtype=pto.i32)

    for i in range(4):
        m = m
        m = pto.cast(i, pto.i32)
    _ = m

try:
    test3_last_value.compile().mlir_text()
    print("compiled successfully.")
except Exception as e:
    print(f"Error compiling: {e}")
```

### 04 for_(...).carray(...)写法

```python
@pto.jit(target="a5", ast_rewrite=True)
def test4_last_value():
    m = pto.const(0, dtype=pto.i32)

    loop = pto.for_(0, 4, step=1).carry(m=m)
    with loop:
        i = loop.iv
        loop.update(m=pto.cast(i, pto.i32))
    _ = loop.final("m")

try:
    test4_last_value.compile().mlir_text()
    print("compiled successfully.")
except Exception as e:
    print(f"Error compiling: {e}")
```
